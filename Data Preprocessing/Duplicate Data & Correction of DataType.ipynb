# Duplicate Values
* Duplicate data can mislead analysis, introduce bias, and affect machine learning model performance.
* Identifying and handling duplicate values is a crucial step in data cleaning during Exploratory Data Analysis (EDA).
* Keeping duplicate data can lead to incorrect conclusions and distorted results.  

## Checking for Duplicate Values
`df.duplicated()` is a Pandas function that checks each row to check for duplicate of a previous row. And Return series of Boolian values  :
> True → row is a duplicate  
> False → row is unique

```
DataFrame.duplicated(subset=None, keep='')
```
**subset** *(default = None, optional)* → Columns to check for duplicates. If None, all columns are used. Example: `subset=['A', 'B']`  
**keep** *(default = 'first', optional)* → Decides which duplicates to mark as False (unique) and which as True (duplicate):

> 'first' → Keep first occurrence, mark others as True.  
> 'last' → Keep last occurrence, mark earlier ones as True.  
> False → Mark all duplicates as True.

````python
# Rerurn series of boolian values
AB_NYC.duplicated()
````

Use `df.duplicated().sum` to get number of duplicated rows

````python
# Check for number of duplicate data
AB_NYC.duplicated().sum()
````

````python
# Finding Duplicates in Multiple Columns
AB_NYC.duplicated(["id", "name"]).sum()
````

````python
# as 'keep=False' mark all duplicate as true
AB_NYC.duplicated(keep=False).sum()
````

## Removing Duplicates
`df = df.drop_duplicates()` is used to remove duplicate rows from a DataFrame, keeping only the first or last occurrence based on your preference.  
Syntax:  
```
DataFrame.drop_duplicates(subset=None, keep='', inplace=False, ignore_index=False)
```
**subset** *(default = None, optional)* → The column label to check for duplicates.  
**keep** *(default = 'first', optional)* → Which duplicate to keep  
**inplace** *(default = False, optional)* → If `True`, modifies the orignal DataFrame and returns None; if `False`, returns a new DataFrame.  
**ignore_index** *(default = False, optional)* → If `True`, resets the index after removing duplicates.

````python
# in this case we dont need to change the indexing
df = AB_NYC.drop_duplicates()
````

````python
AB_NYC.drop_duplicates(keep='last')
````

````python
AB_NYC.drop_duplicates(keep='last',ignore_index=True,inplace=True)
AB_NYC
````

## Handling Duplicates in a Specific Column

````python
# Creating the DataFrame
cities = ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix','Los Angeles', 'Chicago']
rankings = [1, 2, 3, 4, 5,6,7]
scores = [9.8, 9.5, 9.2, 8.7, 8.5, 8.2, 8.1]

data = {'City': cities, 'Ranking': rankings, 'Score': scores}
df = pd.DataFrame(data)
df
````

````python
# Checking for duplicate in whole DataFrame
df.duplicated().sum()
````

````python
# Checking for duplicate in "City" Column
df.duplicated(subset=["City"]).sum()
````

````python
# Removing duplicate while keeping the first occurrence
df.drop_duplicates("City", keep="first")
````

````python
# Removing duplicate while keeping the last occurrence
df.drop_duplicates("City", keep="last")
# It disturb the indexing
````

````python
# Arranged the indexing and Saved in orignal DataFrame
df.drop_duplicates("City", keep='last',ignore_index=True,inplace=True)
df
````
