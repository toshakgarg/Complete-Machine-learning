# Null Value Handling

Different Methods of Dealing With Missing Data
1. **Removing Data**  
  * Removing **columns** with too many missing values.  
  * Removing **rows** that contain missing values.
2. **Filling Missing Values (Imputation)**
  * **For numerical data:** Use **mean, median, or mode.**
  * **For categorical data:** Use **mode** or introduce a **new category.**
3. **Advanced Imputation Techniques**
  * **Interpolation** to estimate missing values.
  * **Predictive models** to estimate missing values based on other variables.

# B) Filling Missing Values (Imputation)
Instead of deleting data, we can replace missing values using **Imputation.**  
**SimpleImputer** is a scikit-learn class which is helpful in handling the missing data in the predictive model dataset. It replaces the NaN values with a specified placeholder.  
It is implemented by the use of the SimpleImputer() method which takes the following arguments :

* **missing_values :** The missing_values placeholder which has to be imputed. By default is NaN  
* **strategy :** The data which will replace the NaN values from the dataset. The strategy argument can take the values – ‘mean'(default), ‘median’, ‘most_frequent’ and ‘constant’.  
* **fill_value :** The constant value to be given to the NaN data using the constant strategy.  

Syntax for imputation
```
import numpy as np
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(missing_values=np.nan, strategy='', fill_value='')
imputer.fit(df.iloc[ : , : ])
df.iloc[ : , : ] = imputer.transform(df.iloc[ : , : ])
```
Syntax for in-build pandas function
```
df[COLUMN_NAME].fillna("N/A or PLACEHOLDER", inplace=True)
```

### 1) Handling Numerical Data
This is done using imputation techniques, such as:
* **Replace with "N/A" or any placeholder :** Can be anything.
* **Mean :** Good for normally distributed data.
* **Median :** Best for data with outliers.
* **Most Frequent :** Works well for categorical-style numeric data.

Placeholder:
```
df[COLUMN_NAME].fillna("N/A or PLACEHOLDER", inplace=True)
placeholder maybe anything(mean, median, mode, or any specific value, etc.)
```
Mean
```
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
```
Median
```
imputer = SimpleImputer(missing_values=np.nan, strategy='median')
```
Mode/ Most frequent
```
imputer = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
```

````python
# Using fillna in-built pandas function
mean = AB_Data['reviews_per_month'].mean()
AB_Data['reviews_per_month'] = AB_Data['reviews_per_month'].fillna(mean)
````

Using Imputation

````python
#Checking slicing for imputer
sample.iloc[ : , 1:3]
````

````python
from sklearn.impute import SimpleImputer

# Create an imputer to fill missing values with the most frequent value in each column
imputer = SimpleImputer(missing_values=np.nan, strategy='most_frequent')

# Fit the imputer on columns 1 and 2 (index starts from 0)
imputer.fit(sample.iloc[:, 1:3])

# Replacing the missing values
sample.iloc[:, 1:3] = imputer.transform(sample.iloc[:, 1:3])

display(sample)
````

### 2) Handling Categorical Data
In Categorical Data we can't use mean, median. So we choose the most_frequent as stratergy for filling up the null values.

* **Replace with "N/A" or any placeholder :** Can be domain-specific.

* **Use Mode/most_frequent (default) :** Good for small datasets with repeating categories.

* **Use "Unknown" OR "New_Category" :**  Useful when missing values need to be explicitly labeled.

Placeholders
```
df[COLUMN_NAME].fillna("N/A or placeholder", inplace=True)
```
      placeholder maybe anything(mean, median, mode, specific value, etc.)

Most frequent
```
imputer = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
```
Unknown or New category
```
imputer = SimpleImputer(missing_values=np.nan, strategy='constant', fill_value="Unknown")
```

````python
MFV = AB_Data["last_review"].value_counts().index[0] #it take first data from count dataframe
AB_Data["last_review"] = AB_Data["last_review"].fillna(MFV)
````
