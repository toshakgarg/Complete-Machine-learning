## Numeric
1. **int:** Integers represent whole numbers without any decimal point. It can be both positive and negative numbers.
2. **float:** Float type represent real numbers and can have decimal points.  
3. **Complex Numbers:** Complex numbers have a real and an imaginary part. It is specified as (real part) + (imaginary part)j.

````python
# Integer Datatype
a = 5
b = -3456
c = 0
print("Type of a: ", type(a))
print("Type of b: ", type(b))
print("Type of c: ", type(c))
````

````python
# Float Datatype
a = 5.0
b = -456.54
print("Type of a: ", type(a))
print("Type of b: ", type(b))
````

````python
# Complex DataTyoe
a = 2 + 4j
b = 3 - 5j
print("Type of a: ", type(a))
print("Type of b: ", type(b))
````
