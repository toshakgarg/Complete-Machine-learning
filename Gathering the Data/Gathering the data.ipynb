## Dtype
Most of the dataset contain unnecessary larger datatype. Or if we require to change the datatype of data while fetching the use dtype and pass dictionary as value.


`````python
# Here in this dataset the last column "Target" is written in float datatype
df=pd.read_csv('aug_train.csv') # Whereas the values in "Target" column is in int type
df
`````

`````python
# Getting the information about dataset before changing the datatype
print(df.info())
print("*"*50,"\n")

# Getting the infornation about dataset after changing the datatype
print(pd.read_csv('aug_train.csv',dtype={'target':int}).info()," \n\n Data set: \n")
df
`````

## Handling Dates
Dates column in dataset generally considered as string, But to deal with dates dataset we can change it using : ```parse_dates=[date]```

---
Note
* If we want to pass multipal column then we can just pass the index of that column as `list`.
* If we have a dataset having dates, Month, Year in different column then we can pass `list of list` for such cases.

`````python
# In this dataset the date column is in oject datatype
IPL=pd.read_csv('IPL Matches 2008-2020.csv')
IPL
`````

`````python
# Geeting the information about our dataset before changing the date datatype
IPL.info()

print("*"*50,"\n")
# Changing the datatype of our date column
IPL=pd.read_csv('IPL Matches 2008-2020.csv',parse_dates=['date'])
# Geeting the information aout our dataset after applying changes
print(IPL.info())
`````

## Convertors
To apply some transformations on our dataset even before you have loaded the dataset we use ``` converoters={'COLUMN_NAME': FUNCTION_NAME}``` and pass dictionary as value.


`````python
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    else:
        return name
`````

`````python
pd.read_csv('IPL Matches 2008-2020.csv',converters={'team1':rename})
`````

## na_values parameter
Some dataset containing `NaN` values which are easier to deal, But some dataset contain placeholder insted of NaN then we use ```na_values``` and pass the list of those placeholders.
