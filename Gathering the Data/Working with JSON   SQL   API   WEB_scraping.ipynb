# Working with API
API (Application programming interface): An API is the set of rules that allow different software or system to talk to each other and share data without acessing the main database.  
Procedure as follow:
* Get the API link
* Generate the API_KEY after Login to site
* Using some set of code acess the data from API

`````python
# Create an empty dataframe
# Adding new data while ignoring new index
# Use for data fetching using API form various API link of same database

df = pd.DataFrame() # Create an Empty dataFrame
print("Before Appending")
display(df)

# Creating temp DataFrame and adding it to existing Dataframe
tempdf = pd.DataFrame({'col1': [1, 2], 'col2': [3, 4]})
df = pd.concat([df, tempdf], ignore_index=True) # To append the data into dataframe

print("After Appending")
display(df)
`````
