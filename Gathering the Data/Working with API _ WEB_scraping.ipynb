## Apending (Ignoring index)

`````python
# Concept of ignoring index in apending
# Used for data fetching using API form various API link of same database

df = pd.DataFrame() # Create an Empty dataFrame
df1 = pd.DataFrame() # Create an Empty dataFrame
tempdf = pd.DataFrame({'col1': [1, 2], 'col2': ["Hello", "World"]}) # Adding the basic content in Dataframe
df = pd.concat([df, tempdf], ignore_index=True) # append the data into dataframe
print("Before Appending")
display(df)

tempdf = pd.DataFrame({'col1': [3, 4], 'col2': ["Toshak", "Here"]})
df1 = pd.concat([df, tempdf]) # append the data into dataframe Without ignoring the index
df = pd.concat([df, tempdf], ignore_index=True) # append the data into dataframe ignoring the index

print("\nAfter Appending")
display(df)
print("\nAfter Appending Without ignoring the index")
display(df1)
`````

`````python
Webpage = requests.get("https://engineering.careers360.com/colleges/ranking?page=1&icn=engineering-ranking_landing-page&ici=data-section_c360-home&sort_by=1").text
Soup = BeautifulSoup(Webpage, 'lxml') # Creating Object to acess methods of class = BeautifulSoup

# Creating list of all card_blocks to fetch data in easier way
Colleges = Soup.find_all("div", class_= 'card_block')
len(Colleges)
`````

`````python
# Append and display length of all the data of each 50 colleges

Name = []       # IIT Madras (IITM)
C_Rating = []   # AAAAA
Ownership = []  # Public/Government
Rating = []     # 4.7/5
Reviews = []    # 127 Reviews

for i in Colleges:
  Name.append(i.find('h3', class_="college_name_heading").text)
  C_Rating.append(i.find_all('span')[1].find_all('p', class_="college-rank-rating")[1].text)
  Ownership.append(i.find_all('span')[3].find('p', class_="college-rank-rating").text)
  Rating.append(i.find_all('span')[5].find('p', class_="college-rank-rating").text)
  Reviews.append(i.find_all('span')[5].find('a').text)

dist = {'Name': Name, 'C_Rating': C_Rating, 'Ownership': Ownership, 'Rating': Rating, 'Reviews': Reviews}
df = pd.DataFrame(dist)
df.head()
df.shape
`````

`````python
# Combining all code in one cell and fetching data from all pages
# There are 26 pages and each pages have 50 colleges
# Final dataset have shape (1300,5)
Final = pd.DataFrame()

for j in range(1,27):# Traverse through each 26 pages
  URL = "https://engineering.careers360.com/colleges/ranking?page={}&icn=engineering-ranking_landing-page&ici=data-section_c360-home&sort_by=1".format(j)
  Request = requests.get(URL).text
  Soup = BeautifulSoup(Webpage, 'lxml') # Creating Object to acess methods of class = BeautifulSoup

  Colleges = Soup.find_all("div", class_= 'card_block') # Creating list of all card_blocks to fetch data in easier way

  # Appending data of each colleges
  Name = []       # IIT Madras (IITM)
  C_Rating = []   # AAAAA
  Ownership = []  # Public/Government
  Rating = []     # 4.7/5
  Reviews = []    # 127 Reviews

  for i in Colleges:
    Name.append(i.find('h3', class_="college_name_heading").text)
    C_Rating.append(i.find_all('span')[1].find_all('p', class_="college-rank-rating")[1].text)
    Ownership.append(i.find_all('span')[3].find('p', class_="college-rank-rating").text)
    Rating.append(i.find_all('span')[5].find('p', class_="college-rank-rating").text)
    Reviews.append(i.find_all('span')[5].find('a').text)

  #Creating and Converting dictionary to Final DataFrame
  dist = {'Name': Name, 'C_Rating': C_Rating, 'Ownership': Ownership, 'Rating': Rating, 'Reviews': Reviews}
  df = pd.DataFrame(dist)
  Final = pd.concat([Final, df], ignore_index=True) #Ignoring index

#Printing the size of College_DataFrame
Final.shape
Final.to_csv('Gathering the Data/College_Data.csv')
`````
