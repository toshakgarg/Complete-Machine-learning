# **Bivariate Analysis**
Bivariate analysis helps in understanding the relationship between two variables in a dataset. In bivariate analysis, we focus on how two variables, either categorical or numerical, relate to each other.

## **Bivariate Visualization Techniques**
* **Bar plots** allow us to compare numerical values (like mean or sum) across different categories.
* **Box plots** show the distribution and spread of numerical data, highlighting the median, quartiles, and potential outliers.
* **Violin plots** combine distribution and box plot features to visualize both the distribution and probability density of numerical data.
* **Distribution plots** offer a way to compare the distribution of a numerical variable across categories.

# **Bivariate Analysis: Combining Numerical and Categorical Data**
Bivariate analysis allows us to investigate how a numerical variable changes in response to different categories within a categorical variable.

### **few Scenarios:**
* Does student performance (numerical) vary based on the type of learning environment (categorical)?
* Are there significant differences in salary (numerical) across different job roles (categorical)?

## **Violin Plot**
Combining Distribution and Box Plot.
A violin plot combines the benefits of a box plot and a kernel density plot. It shows the distribution of data along with its probability density, making it a great tool for understanding the distribution across different categories.

## **1. Scatterplot (Numerical - Numerical)**
`sns.scatterplot()` is used to visualize the relationship between two continuous variables. Each point represents one observation, making it ideal to detect correlations, trends, and outliers.  

🔹 **Insights / When to Use**
* Detects linear or non-linear relationships between two variables.
* Reveals clusters, outliers, and group patterns (via hue/style).
* Useful for both continuous vs continuous and continuous vs categorical (when using style or hue).

Syntax:
```
sns.scatterplot(data=None, x=None, y=None, hue=None, style=None, size=None, palette=None)
```
**🔹 Parameters**
* **data** *(default = None, optional):* Dataset (like a DataFrame) containing variables for x and y.
* **x** *(default = None, optional):* Column name or array for x-axis values.
* **y** *(default = None, optional):* Column name or array for y-axis values.
* **hue** *(default = None, optional):* Grouping variable that adds color based on categories or numeric scale.  
* **style** *(default = None, optional):* Changes marker shape based on a categorical variable.
* **size** *(default = None, optional):* Varies point size based on a numeric or categorical variable.
* **palette** *(default = None, optional):* Color palette for hue variable (e.g., "coolwarm", "viridis").


````python
# Scatterplot for just two columns
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip')
````

````python
# But we can also catogrize them based on colour, style & size using these hue, style, size respective parameters
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', hue = 'sex', style = 'smoker', size = 'size')
````

````python
# But we can also catogrize them based on colour, style & size using these hue, style, size respective parameters
# changing the colour palette to 'viridis'
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', hue = 'sex', style = 'smoker', size = 'size', palette = 'viridis')
````

## **2. Bar Plot (Numerical - Categorical)**
`sns.barplot()` shows the average value of a numeric variable for each category. It can show the central tendency (like the mean or sum) of a numerical variable for each category in a categorical variable.  

**🔹 Insights / When to Use**
* Ideal for comparing group averages or category-wise trends.
* Useful to identify highest/lowest performing categories.
* useful to find distribution of numerical value under a category.
* hue helps in comparing multiple categorical divisions together.
* Bar height → central tendency (mean or median).
* Error bars → variability or uncertainty in data.

Syntax:
```
sns.barplot (data=None, x=None, y=None, hue=None, order=None, hue_order=None, estimator='mean')
```
* **x** (optional, default=None): Variable assigned to x-axis *(categorical if vertical bar plot)*.
* **y** (optional, default=None): Variable assigned to y-axis *(numerical if vertical bar plot)*.
* **data** (optional, default=None): Dataset.
* **hue** (optional, default=None): Additional categorical variable *(grouping within each category)*.
* **order** (optional, default=None): Order of categories for x.
* **hue_order** (optional, default=None): Order of categories for hue.
* **estimator** (optional, default=mean): Function to compute value of bar *(e.g., np.mean, np.median, len)*.

````python
# Plotting a bar plot for 'age' by 'class'
sns.barplot(data=titanic, x="class", y="age")
````

````python
# Plotting a bar plot for 'age' by 'class', with hue 'sex'
sns.barplot(data=titanic, x="class", y="age", hue="sex")
````

````python
# Plotting a bar plot with sum insted of average of 'age' by 'class'
sns.barplot(data=titanic, x="class", y="age", estimator="sum")
````

## **3. Box Plot (Numerical - Categorical)**
`sns.boxplot()` displays the distribution of a numeric variable across categories using five summary statistics — minimum, Q1, median, Q3, and maximum — *with outliers shown as individual points.*  

**🔹 Insights / When to Use**
* Great for detecting outliers and understanding data spread & skewness.
* It’s useful for comparing the distribution of a numerical variable across different categories.
* Median line → central value.
* Box height → interquartile range (IQR).
* Whiskers → data variability.
* Outlier points → unusual data values.
* Use hue to compare multiple categories side-by-side.


Syntax:
```
sns.boxplot(data=None, x=None, y=None, hue=None, palette=None, orient=None, width=0.8)
```
* data (default = None, optional): DataFrame containing the data
* x (default = None, optional): Categorical variable (for grouping).
* y (default = None, optional): Numeric variable whose distribution is shown.
* hue (default = None, optional): Adds sub-categories with different colors.  
* palette (default = None, optional): Color palette for hue (e.g., "coolwarm", "Set2").  
* orient (default = None, optional): "v" for vertical or "h" for horizontal boxes.
* width (default = 0.8, optional): Controls box width (smaller = thinner boxes).

````python
# Plotting a box plot for 'age' by 'sex'
sns.boxplot(x="sex", y="age", data=titanic)
````

````python
# Plotting a box plot for 'age' by 'sex', with hue 'class' and wdth 0.5
sns.boxplot(x="sex", y="age", hue="survived", data=titanic, width= 0.5)
````

## **4. Distplot (Numerical - Categorical)**
### **Distplot: Visualizing *Probability distribution function***
`sns.distplot()` plots the distribution of a numeric variable, combining a histogram and an optional KDE (Kernel Density Estimate) curve. It helps visualize the shape, spread, and skewness of data. Histogram tell us about the exact numeber while Distplot give the probablity of that number` using *Probability distribution function*

🔹 Insights / When to Use
* Ideal for understanding data distribution (normal, skewed, uniform, etc.).
* KDE curve gives insight into density and smoothness.
* Histogram shows frequency counts or probabilities.
* Great for checking normality before applying statistical tests or ML models.
* Here in this graph the Histogram is overlaped by an line called *Probability distribution function*, And it is calculated by *Kernel Density Estimation.*  


Syntax:
```
sns.distplot(a=None, bins=None, hist=True, kde=True, color=None, vertical=False, norm_hist=False)
```

* a (default = None, required): The data (array, list, or Series) to plot.
* bins (default = None, optional): Number of histogram bins (e.g., 10, 20).
* hist (default = True, optional): Whether to draw the histogram.
* kde (default = True, optional): Whether to draw the KDE (smooth density cr* ve).
* vertical (default = False, optional): If True, flips the plot vertically.
* color (default = None, optional): Sets the color of bars and KDE line.
* norm_hist (default = False, optional): If True, normalizes the histogram (so area = 1).

````python
#Ploting Distribution plot for age only
sns.distplot(titanic["age"])
````

````python
# Plotting distribution plots for 'age' by 'survived'
# Ploting graph simultaneously of both survived or not by creating new dataFrame of both.
sns.distplot(titanic[titanic["survived"] == 0]["age"]) # survived = titanic[titanic['survived'] == 1]
sns.distplot(titanic[titanic["survived"] == 1]["age"]) # Not_survived = titanic[titanic['survived'] == 0]
# Plotting distribution plots for 'age' by 'survived'. By creating a DataFrame for both Survived/ Not Survived
````

````python
# Plotting distribution plots for 'age' by 'survived'
sns.distplot(titanic[titanic["survived"] == 0]["age"], hist=False)
sns.distplot(titanic[titanic["survived"] == 1]["age"], hist=False)
````

````python
# Plotting distribution plots for 'age' by 'survived' with changed color
sns.distplot(titanic[titanic["survived"] == 0]["age"], hist=False, color = 'blue')
sns.distplot(titanic[titanic["survived"] == 1]["age"], hist=False, color = 'red')
````

## **5. HeatMap (Categorical - Categorical)**

````python
# To determine no of perple survived in different classes
pd.crosstab(titanic['pclass'], titanic['survived'])
````

````python
# Ploting the graph of no of perple survived in different classes
sns.heatmap(pd.crosstab(titanic['pclass'], titanic['survived']))
# Here we can get only number of people died in respective class
````

````python
# Better measure of determination is percentage
display(titanic.groupby('pclass')['survived'].mean() * 100)
(titanic.groupby('pclass')['survived'].mean() * 100).plot(kind='bar')
````

## **6. ClusterMap (Categorical - Categorical)**

````python
display(pd.crosstab(titanic['parch'], titanic['survived']))
sns.clustermap(pd.crosstab(titanic['parch'], titanic['survived']))
# Tells the clossness between the values
# Also capture the hierarchical relation between data using dendograms
````
