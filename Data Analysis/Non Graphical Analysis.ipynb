## 3. Analyzing Numerical Data by category
To analyze numerical data using categories, we use binning. Binning allows you to segment the data into meaningful intervals, making it easier to identify age groups, or to analyze patterns within specific ranges.

Syntax:
```
pd.value_counts(normalize: bool=False, sort: bool=True, ascending: bool=False, bins=None)
```
> Normalize →  (default = False) (optional) If True then return relative frequencies of the unique values. (Percentage)  
> sort →  (default = True) (optional) Sorts results by counts if True.  
> ascending →  (default = False) (optional) If True, sorts smallest to largest.  
> bins → (default = None) (optional) Groups values into intervals (for numerical data).  
> dropna → (default = True) (optional) Ignores NaN values if True.

`bins=n` : This divides the column into `n` equal bins and counts occurrences in each range.  
We can define custom bins using `bins = [-10, 0, 50, 100...]` and `bins = bins`.

````python
AB_NYC["price"].value_counts(bins=5)
````

````python
bins = [-10, 0, 50, 100, 200, 500, 800, 2000, 4000, 10000]
AB_NYC["price"].value_counts(bins=bins, sort= False)
````

## 4. Measures of Central Tendency
It help identify a single value that best represents the center of a dataset. They give an overall idea of where most data points lie and are useful for summarizing and comparing datasets.

* **Mean** – The average of all values; shows the dataset’s overall balance point.  
Syntax: `df['COLUMN_NAME'].mean()`  

* **Median** – The middle value when data is sorted; less affected by outliers.  
Syntax: `df['COLUMN_NAME'].median()`  

* **Standard Deviation** – Measures how spread out the data is from the mean.  
Syntax: `df['COLUMN_NAME'].std()`

* **Skewness** – Shows whether the data leans left, right, or is symmetric.  
Syntax: `df['COLUMN_NAME'].skew()`

* **Kurtosis** – Indicates whether the data has heavy or light tails compared to a normal curve.  
Syntax: `df['COLUMN_NAME'].kurt()`  

````python
mean = AB_NYC["price"].mean()
median = AB_NYC["price"].median()
std = AB_NYC["price"].std()
skew = AB_NYC["price"].skew()
kurt = AB_NYC["price"].kurt()
print(f"Mean:", mean,
      "\nMedian:", median,
      "\nStandard Deviation:", std,
      "\nSkewness:", skew,
      "\nKurtosis:", kurt
      )
````

## 5. Correlation Analysis
Correlation helps determine if two numerical variables are related.

**Correlation:** Correlation is a statistical measure that quantifies the degree of association or relationship between two variables. It is typically used for bivariate analysis but can also be relevant in univariate analysis to explore potential relationships that might exist between different subsets or aspects of a single variable.  
`df.corr()` returns a correlation matrix in the form of a new DataFrame. Where:
* Rows and columns represent the numerical columns of the original DataFrame.
* Values are the correlation coefficients between each pair of columns.
* The diagonal values are always 1.0 (each column perfectly correlates with itself).

````python
AB_NYC.corr(numeric_only=True)
````
